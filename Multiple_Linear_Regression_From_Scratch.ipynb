##MATHEMATICAL INTUTITON OF MULTIPLE LINEAR REGRESSION ALGORITHM FROM SCRATCH

## Phase 1: The Real-World Goal

Imagine we are a college placement cell. We want to predict a student’s **Placement Package ($Y$)** based on two factors: their **CGPA ($X_1$)** and their **IQ ($X_2$)**.

We collect data from 3 past students to build our prediction model:

| Student | CGPA ($X_1$) | IQ ($X_2$) | Placement Package ($Y$) |
| --- | --- | --- | --- |
| **Student 1** | $8$ | $80$ | $8$ Lakhs |
| **Student 2** | $7$ | $70$ | $7$ Lakhs |
| **Student 3** | $5$ | $120$ | $15$ Lakhs |

---

## Phase 2: The Math for One Student (Algebra)

If we want to predict the package ($\hat{Y}$) for any single student, we use a straight-line equation.

Because we have two inputs (CGPA and IQ), our equation looks like this:

$$\hat{Y} = \beta_0 + \beta_1(CGPA) + \beta_2(IQ)$$

Where:

* $\hat{Y}$ is our **prediction**.
* $\beta_0$ is the **intercept** (the starting package if CGPA and IQ were zero).
* $\beta_1$ is the **weight/importance** of CGPA.
* $\beta_2$ is the **weight/importance** of IQ.

If we write this equation out for our three students using their actual data, we get a **system of equations**:

1. **Student 1:** $\hat{Y}_1 = \beta_0 + \beta_1(8) + \beta_2(80)$
2. **Student 2:** $\hat{Y}_2 = \beta_0 + \beta_1(7) + \beta_2(70)$
3. **Student 3:** $\hat{Y}_3 = \beta_0 + \beta_1(5) + \beta_2(120)$

---

## Phase 3: The Problem with Scale

Writing three equations by hand is easy. But what if we have **10,000 students** and **50 different features** (like internship experience, projects, test scores)?

Writing out 10,000 algebraic equations is impossible. Computers solve this by grouping all these numbers into **Matrices**.

---

## Phase 4: Entering the Matrix (The 3 Blocks)

To turn our algebra into matrix math, we pack our numbers into three separate boxes:

### 1. The Target Vector ($Y$ or $\hat{Y}$)

This is an $n \times 1$ column vector (where $n$ is the number of students). It simply stacks all our placement packages vertically:

$$Y = \begin{bmatrix} Y_1 \\ Y_2 \\ Y_3 \end{bmatrix} = \begin{bmatrix} 8 \\ 7 \\ 15 \end{bmatrix}$$

### 2. The Coefficients / Weights Vector ($\beta$)

This is a column vector containing the unknown variables ($\beta_0, \beta_1, \beta_2$) that our model needs to figure out:

$$\beta = \begin{bmatrix} \beta_0 \\ \beta_1 \\ \beta_2 \end{bmatrix}$$

### 3. The Data Matrix ($X$)

This matrix contains our input data.

> **The Intercept Trick:** Notice that the intercept $\beta_0$ in our algebraic equation is alone—it doesn't have an $X$ variable attached. In matrix multiplication, every term needs a partner. Since $\beta_0 \times 1 = \beta_0$, we trick the matrix by adding a **column of entirely $1$s** as our very first column:

$$X = \begin{bmatrix} 1 & 8 & 80 \\ 1 & 7 & 70 \\ 1 & 5 & 120 \end{bmatrix}$$

---

## Phase 5: How They Fit Together ($\hat{Y} = X\beta$)

Now we can write the entire system of equations in one incredibly short formula:

$$\hat{Y} = X\beta$$

$$\begin{bmatrix} \hat{Y}_1 \\ \hat{Y}_2 \\ \hat{Y}_3 \end{bmatrix} = \begin{bmatrix} 1 & 8 & 80 \\ 1 & 7 & 70 \\ 1 & 5 & 120 \end{bmatrix} \begin{bmatrix} \beta_0 \\ \beta_1 \\ \beta_2 \end{bmatrix}$$

### Why does this work?

In matrix multiplication, you multiply **rows** by **columns**.

If we multiply the first row of $X$ by the column of $\beta$:


$$\hat{Y}_1 = (1 \times \beta_0) + (8 \times \beta_1) + (80 \times \beta_2)$$

$$\hat{Y}_1 = \beta_0 + 8\beta_1 + 80\beta_2$$

This matches our algebraic equation for Student 1 perfectly!

---

## Phase 6: Finding the Best Parameters (The Final OLS Formula)

In real life, we don't know what $\beta_0, \beta_1,$ and $\beta_2$ are. We must calculate them so that our predictions ($\hat{Y}$) are as close as possible to the actual results ($Y$).

The difference between actual values and predicted values is the **error ($e$)**:

$$e = Y - \hat{Y}$$

To find the absolute best values for $\beta$ that minimize this error, we do some calculus using the matrices.

By setting the derivative of the squared error to zero, we arrive at the famous **Ordinary Least Squares (OLS) closed-form solution**:

$$\beta = (X^T X)^{-1} X^T Y$$

### Breaking down the OLS steps:

1. **$X^T$ (Transpose of $X$):** Flips the data matrix rows into columns.
2. **$X^T X$:** Multiplies the flipped matrix by the original matrix to get a square matrix.
3. **$(X^T X)^{-1}$ (Inverse):** The matrix equivalent of dividing (like calculating $\frac{1}{x}$).
4. **$X^T Y$:** Multiplies the flipped data by the actual placement packages.

By running this single calculation, your computer finds the exact optimal values for $\beta_0, \beta_1,$ and $\beta_2$ instantly, giving you your finished predictive model.

```
class selfMLR:
  def __init__(self):
    self.M = None
    self.B = None

  def fit(self,X_train,Y_train):
    X_train = np.insert(X_train,0,1,axis=1)

    self.betas = np.linalg.inv(np.dot(X_train.T,X_train)).dot(X_train.T).dot(Y_train)
    self.M = self.betas[1:]
    self.B = self.betas[0]

  def predict(self,X_test):
    ## Y = X * B (Y = mx + c )(Y = M * X + B)
    self.y_predicted = np.dot(X_test,self.M) + self.B
    return self.y_predicted
```

```
X,y = load_diabetes(return_X_y=True)
X.shape
```

```
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(X,y,test_size=0.2, random_state = 2)
```

```
mlr = selfMLR()
mlr.fit(X_train,Y_train)
y_pred = mlr.predict(X_test)
```

```
from sklearn.metrics import r2_score
r2_score(Y_test,y_pred)
```

```
from sklearn.linear_model import LinearRegression
mlr = LinearRegression()
mlr.fit(X_train,Y_train)
y_pred = mlr.predict(X_test)
```

```
r2_score(Y_test,y_pred)
```
